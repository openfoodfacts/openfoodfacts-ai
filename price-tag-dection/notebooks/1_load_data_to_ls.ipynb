```python
import requests
import gzip
import shutil
import pandas as pd
import json
```

```python
proof_url = "https://prices.openfoodfacts.org/data/proofs.jsonl.gz"
data_path = "/Users/baslad01/data_dump/openfoodfacts"
```

```python
image_prefix_url = "https://prices.openfoodfacts.org/img/"

sampling = None

# Download the file
r = requests.get(proof_url, stream=True)
with open(data_path + "proofs.jsonl.gz", "wb") as file:
    for chunk in r.iter_content(chunk_size=1024):
        file.write(chunk)


# Unzip the file
with gzip.open(data_path + "/proofs.jsonl.gz", 'rb') as f_in:
    with open(data_path + "/proofs.jsonl", 'wb') as f_out:
        shutil.copyfileobj(f_in, f_out)
```

```python
# Read the file
df = pd.read_json(data_path + "/proofs.jsonl", lines=True)
df.head()
```

```python
# Filter where type is PRICE_TAG
df = df[df["type"] == "PRICE_TAG"]
df.head()
```

```python
# Select a random sample of 1000 rows
if sampling:
    df = df.sample(sampling)
    df.head()
```

```python
# Extract the URLs from the DataFrame
urls = df['file_path'].tolist()
# ids = df['id'].tolist()
df['image_id'] = df['file_path'].apply(lambda x: x.split('/')[-1].split('.')[0])
image_ids = df['image_id'].tolist()

# Format the URLs into the desired JSON structure
# formatted_data = [{"id": id,"data": {"image_url": image_prefix_url + url}} for url, id in zip(urls, ids)]
formatted_data = [{"data": {"image_url": image_prefix_url + url, "image_id": image_id}} for url, image_id in zip(urls, image_ids)]

# Export the JSON structure to a file
output_path = f'{data_path}/formatted_urls.json'
with open(output_path, 'w') as json_file:
    json.dump(formatted_data, json_file, indent=2)

print(f"Formatted data saved to {output_path}")
```
