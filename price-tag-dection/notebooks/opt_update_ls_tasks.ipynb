```python
from tqdm import tqdm
import requests
import os
import json
```

```python
# Read the data from json file
data_path = "/Users/baslad01/data_dump/openfoodfacts"
file_name = "formatted_urls.json"
file_path = os.path.join(data_path, file_name)

with open(file_path, 'r') as file:
    data = json.load(file)

```

```python
project_id = 56
ls_url = f"https://annotate.openfoodfacts.org"
get_tasks_url = f"{ls_url}/api/tasks/"
get_projects_url = f"{ls_url}/api/projects/"

headers = {
    "Authorization": f"Token {os.getenv('LABEL_STUDIO_KEY')}"
}

all_tasks = []
page = 1

while True:
    response = requests.get(get_tasks_url, headers=headers, params={"project": project_id, "page": page})
    data = response.json()
    tasks = data.get('tasks', [])
    if not tasks:
        break
    all_tasks.extend(tasks)
    page += 1
```

```python
# Get the tasks ids
task_ids = [task['id'] for task in all_tasks]
```

```python
# for each task id, get the image_url and add the image_id field which is the last part of the image_url
task_ids_s = [35128]
for task_id in tqdm(task_ids, desc="Processing tasks"):
    task_url = f"{get_tasks_url}{task_id}/"
    response = requests.get(task_url, headers=headers)
    task = response.json()
    image_url = task['data']['image_url']
    image_id = image_url.split("/")[-1].split('.')[0]
    task['data']['image_id'] = image_id
    task['updated_by'] =  47 # Replace with the correct pk value
    task['file_upload'] = 1
    response = requests.put(task_url, headers=headers, json=task)
    # print(response.json())
```
